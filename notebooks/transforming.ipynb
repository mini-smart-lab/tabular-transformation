```python
import pandas as pd
import numpy as np  
import os
import yaml

import sys
sys.path.append('../')
np.random.seed(42)

from pathlib import Path
from transformation.log import LogTransformation
from transformation.sqrt import SquareRootTransformation
from transformation.square import SquareTransformation
```

```python
notebook_dir = os.path.dirname(os.path.abspath("__file__"))
parent_dir = Path(notebook_dir).parent
data_dir = parent_dir / "data"
print(f"There are {len(os.listdir(data_dir))} datasets.")
```

```python
operators = [SquareTransformation()]

folders = os.listdir(data_dir)
for f in folders:
    path = data_dir / f
    train_before = path / "train_before.csv"
    train_after = path / "train_after.csv"
    config_path = path / "config.yaml"
    df = pd.read_csv(train_before)
    cols = df.columns.to_list()
    col_to_transform = np.random.choice(cols, 1)
    operator_index = np.random.choice(range(len(operators)), 1)[0]
    operator = operators[operator_index]

    df[col_to_transform] = operator.transform(df[col_to_transform])
    df.to_csv(train_after, index= False)
    config = {
        "columns" : ",".join(col_to_transform),
        "operator" : operator.__str__(),
    }
    with open(config_path, "w") as file:
        yaml.dump(config, file)

print("Done transformation")
```
