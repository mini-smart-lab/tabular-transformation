```python
import os
import pandas as pd
import numpy as np

np.random.seed(42)

notebook_dir = os.path.dirname(os.path.abspath('__file__'))
data_path = os.path.join(notebook_dir, 'processed_data.csv')
df = pd.read_csv(data_path)
df.head()
```

```python
N_SAMPLES = 10000
n_pairs = len(df) // N_SAMPLES
print(f"Number of pairs: {n_pairs}")
```

```python
from pathlib import Path

current_path = Path(notebook_dir)
parrent_path = current_path.parent
path_to_save = parrent_path / "data"

os.makedirs(path_to_save, exist_ok= True)

for i in range(n_pairs):
    idx = np.random.choice(len(df), N_SAMPLES)
    sample_df = df.iloc[idx]
    p = path_to_save / f"{i}"
    os.makedirs(p, exist_ok= True)
    sample_df.to_csv(p / "train_before.csv")

print("Done sampling data.")
```
