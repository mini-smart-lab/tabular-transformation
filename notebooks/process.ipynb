```python
import os
import pandas as pd

# Lấy thư mục hiện tại chứa notebook
notebook_dir = os.path.dirname(os.path.abspath('__file__'))

# Trỏ đến tệp CSV
train_path = os.path.join(notebook_dir, 'data.csv')
train_df = pd.read_csv(train_path)
```

```python
train_columns = train_df.columns.to_list()
train_columns[1:] = [f"col_{i}" for i in range(1, len(train_columns))]
train_columns
```

```python
train_df.columns = train_columns
train_df.head()
```

```python
for col in train_df.columns.to_list():
    if train_df[col].dtype in ["object", "category", "str"]:
        codes, _ = pd.factorize(train_df[col])
        train_df[col] = codes

train_df.head()
```

```python
output_path = os.path.join(notebook_dir, 'processed_data.csv')
train_df.to_csv(output_path, index = False)
print(f"Saved processed dataframe to {output_path}")
```
